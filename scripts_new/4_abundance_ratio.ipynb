#### **6. Visualizing doubling speed**

```python
import math
r = math.log(2)
```

```python
# Abundances ratio
scenarios = {
    "10%": sol_TyF_10.fluxes,
    "30%": sol_TyF_30.fluxes,
    "50%": sol_TyF_50.fluxes,
    "70%": sol_TyF_70.fluxes,
    "90%": sol_TyF_90.fluxes,
}

#
data = {
    "Abundance": [],
    "CanA": [],
    "CanB": [],
    "CanC": [],
    "Fusobacterium": []
}

# Stimate doubling speed 
for label, fluxes in scenarios.items():
    data["Abundance"].append(label)

    # Cancer
    flux_A = fluxes.loc["CanA", "biomass_reaction"]
    flux_B = fluxes.loc["CanB", "biomass_reaction"]
    flux_C = fluxes.loc["CanC", "biomass_reaction"]

    # Fusobacterium
    flux_F = fluxes.loc["Fusobacterium", "EX_biomass_e"]

    # Doubling speed
    data["CanA"].append(r / flux_A if flux_A > 0 else None)
    data["CanB"].append(r / flux_B if flux_B > 0 else None)
    data["CanC"].append(r / flux_C if flux_C > 0 else None)
    data["Fusobacterium"].append(r / flux_F if flux_F > 0 else None)

# Make DataFrame
df_doubling = pd.DataFrame(data)

# 
df_melted = df_doubling.melt(id_vars="Abundance", var_name="member", value_name="Doubling_speed")

# Visualizing
plt.figure(figsize=(10, 6))

for model in df_melted["member"].unique():
    subset = df_melted[df_melted["member"] == model]
    plt.plot(subset["Abundance"], subset["Doubling_speed"], marker='o', label=model)

#plt.title("Doubling speed", fontsize=14)
plt.xlabel("Abundance of Fusobacterium")
plt.ylabel("Doubling speed (hours)")
plt.legend(title="member")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("plots/tiempo_duplicacion_abundancias.png", dpi=300, bbox_inches="tight")
plt.show()
```
